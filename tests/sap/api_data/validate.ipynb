```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('validate.csv')
df
```

```python
fig, ax = plt.subplots(figsize = (16,6))
ax.scatter(x = df.energy_rating_scheme, y = df.energy_rating_er)
ax.set_xlabel('Original SAP calculation'), ax.set_ylabel('sap10calcs')
plt.show()
```

```python
fig, ax = plt.subplots(figsize = (16,6))
ax.hist(df.energy_rating_er-df.energy_rating_scheme, bins = [x-0.5 for x in range(-100,100)])
ax.set_xlim(-5,5)
plt.show()
```

```python
sum(abs(df.energy_rating_er-df.energy_rating_scheme) == 0) / len(df) * 100
```

```python
sum(abs(df.energy_rating_er-df.energy_rating_scheme) <=1 ) / len(df) * 100
```

```python
sum(abs(df.energy_rating_er-df.energy_rating_scheme) <=2 ) / len(df) * 100
```

```python
fig, ax = plt.subplots(figsize = (16,6))
ax.scatter(x = df.DER_scheme, y = df.DER_er)
ax.set_xlabel('DER - Original SAP calculation'), ax.set_ylabel('DER - sap10calcs')
plt.show()
```

```python
fig, ax = plt.subplots(figsize = (16,6))
ax.hist((df.DER_er-df.DER_scheme)/df.DER_scheme, bins = 1000)
ax.set_xlim(-0.5,0.5)
plt.show()
```

```python
sum(abs((df.DER_er-df.DER_scheme)/df.DER_scheme) <= 0.25) / len(df) * 100
```

```python
fig, ax = plt.subplots(figsize = (16,6))
ax.scatter(x = df.PDER_scheme, y = df.PDER_er)
ax.set_xlabel('PDER - Original SAP calculation'), ax.set_ylabel('PDER - sap10calcs')
plt.show()
```

```python
fig, ax = plt.subplots(figsize = (16,6))
ax.hist((df.PDER_er-df.DER_scheme)/df.PDER_scheme, bins = 1000)
ax.set_xlim(0,2)
plt.show()
```

```python
sum(abs((df.PDER_er-df.PDER_scheme)/df.PDER_scheme) <= 1) / len(df) * 100
```
